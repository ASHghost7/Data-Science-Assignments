## 1. Introduction to Pandas

Pandas is a powerful data manipulation and analysis library for Python. It provides data structures and functions needed to work with structured data.

```python
import pandas as pd
import numpy as np

# Check pandas version
print(f"Pandas version: {pd.__version__}")
```

## 2. Series and DataFrames

### Series - 1D labeled array

```python
# Creating a Series
data = [10, 20, 30, 40, 50]
series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])
print("Series:")
print(series)
print(f"\nData type: {series.dtype}")
print(f"Index: {series.index.tolist()}")
```

### DataFrame - 2D labeled data structure

```python
# Creating a DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'London', 'Tokyo', 'Paris'],
    'Salary': [50000, 60000, 70000, 55000]
}

df = pd.DataFrame(data)
print("DataFrame:")
print(df)
print(f"\nShape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
```

## 3. Basic DataFrame Operations

```python
# Basic information about the DataFrame
print("DataFrame Info:")
print(df.info())
print("\nDescriptive Statistics:")
print(df.describe())
print("\nFirst 2 rows:")
print(df.head(2))
```

## 4. Data Selection and Indexing

```python
# Selecting columns
print("Names column:")
print(df['Name'])
print("\nMultiple columns:")
print(df[['Name', 'Age']])

# Selecting rows
print("\nFirst row:")
print(df.iloc[0])
print("\nFiltering data (Age > 28):")
print(df[df['Age'] > 28])
```

## 5. Data Cleaning

```python
# Create a DataFrame with missing values
data_with_nulls = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, 5]
}

df_nulls = pd.DataFrame(data_with_nulls)
print("DataFrame with missing values:")
print(df_nulls)

# Check for missing values
print("\nMissing values count:")
print(df_nulls.isnull().sum())

# Fill missing values
df_filled = df_nulls.fillna(df_nulls.mean())
print("\nAfter filling with mean:")
print(df_filled)
```

## 6. Grouping and Aggregation

```python
# Create a larger dataset for grouping
sales_data = {
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Region': ['North', 'North', 'South', 'South', 'North', 'South'],
    'Sales': [100, 150, 200, 120, 180, 90],
    'Profit': [20, 30, 40, 24, 36, 18]
}

sales_df = pd.DataFrame(sales_data)
print("Sales Data:")
print(sales_df)

# Group by Product
print("\nGrouped by Product:")
product_group = sales_df.groupby('Product').sum()
print(product_group)

# Multiple aggregations
print("\nMultiple aggregations:")
agg_result = sales_df.groupby('Region').agg({
    'Sales': ['sum', 'mean'],
    'Profit': ['sum', 'max']
})
print(agg_result)
```

## 7. Practice Problems

### Problem 1: Student Grades Analysis

```python
# Create student data
students = {
    'Student': ['John', 'Emma', 'Michael', 'Sarah', 'David'],
    'Math': [85, 92, 78, 96, 89],
    'Science': [88, 94, 82, 90, 87],
    'English': [82, 89, 85, 93, 91],
    'Grade': ['A', 'A', 'B', 'A', 'A']
}

students_df = pd.DataFrame(students)
print("Student Data:")
print(students_df)

# Calculate average score for each student
students_df['Average'] = students_df[['Math', 'Science', 'English']].mean(axis=1)
print("\nWith Average:")
print(students_df)

# Find students with average > 85
high_performers = students_df[students_df['Average'] > 85]
print("\nHigh Performers (Average > 85):")
print(high_performers[['Student', 'Average']])
```

### Problem 2: Sales Data Analysis

```python
# Create monthly sales data
monthly_sales = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Product_A': [1000, 1200, 1100, 1300, 1250, 1400],
    'Product_B': [800, 900, 850, 950, 1000, 1100],
    'Product_C': [600, 650, 700, 750, 800, 850]
}

monthly_df = pd.DataFrame(monthly_sales)
print("Monthly Sales Data:")
print(monthly_df)

# Calculate total sales per month
monthly_df['Total_Sales'] = monthly_df[['Product_A', 'Product_B', 'Product_C']].sum(axis=1)
print("\nWith Total Sales:")
print(monthly_df)

# Find the month with highest sales
best_month = monthly_df.loc[monthly_df['Total_Sales'].idxmax()]
print(f"\nBest performing month: {best_month['Month']} with sales: {best_month['Total_Sales']}")

# Calculate growth rate
monthly_df['Growth_Rate'] = monthly_df['Total_Sales'].pct_change() * 100
print("\nWith Growth Rate:")
print(monthly_df[['Month', 'Total_Sales', 'Growth_Rate']])
```

## 8. Quick Reference

### Common Pandas Operations:

- `pd.read_csv('file.csv')` - Read CSV file
- `df.head(n)` - Show first n rows
- `df.tail(n)` - Show last n rows
- `df.info()` - DataFrame information
- `df.describe()` - Descriptive statistics
- `df.shape` - DataFrame dimensions
- `df.columns` - Column names
- `df.isnull().sum()` - Count missing values
- `df.fillna(value)` - Fill missing values
- `df.dropna()` - Drop rows with missing values
- `df.groupby('column').agg()` - Group and aggregate
- `df.sort_values('column')` - Sort by column
- `df.drop_duplicates()` - Remove duplicates
